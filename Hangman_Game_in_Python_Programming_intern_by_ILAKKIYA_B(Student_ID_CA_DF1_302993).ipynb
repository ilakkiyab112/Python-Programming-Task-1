```python
import random

WORDS = [
    "python",
    "computer",
    "programming",
    "developer",
    "security"
]

MAX_ATTEMPTS = 6


def display_word(word, guessed_letters):
    result = []

    for letter in word:
        if letter in guessed_letters:
            result.append(letter.upper())
        else:
            result.append("_")

    return " ".join(result)


def display_status(word, guessed_letters, wrong_guesses):
    print("\n" + "=" * 45)
    print("HANGMAN GAME")
    print("=" * 45)

    print("\nWord:")
    print(display_word(word, guessed_letters))

    print(f"\nWrong guesses: {wrong_guesses}/{MAX_ATTEMPTS}")

    if guessed_letters:
        print(
            "Guessed letters:",
            " ".join(
                sorted(letter.upper() for letter in guessed_letters)
            )
        )


def play_game():

    word = random.choice(WORDS)

    guessed_letters = set()
    wrong_guesses = 0

    print("\nWelcome to the Hangman Game!")
    print("Guess the hidden word one letter at a time.")
    print(f"You have {MAX_ATTEMPTS} incorrect guesses.")

    while wrong_guesses < MAX_ATTEMPTS:

        display_status(
            word,
            guessed_letters,
            wrong_guesses
        )

        # Check whether player has won
        if all(letter in guessed_letters for letter in word):
            print("\nCONGRATULATIONS!")
            print(f"You guessed the word: {word.upper()}")
            return

        guess = input("\nEnter a letter: ").strip().lower()

        # Validate input
        if len(guess) != 1 or not guess.isalpha():
            print("Please enter exactly one alphabet letter.")
            continue

        # Check repeated guess
        if guess in guessed_letters:
            print("You already guessed that letter.")
            continue

        guessed_letters.add(guess)

        # Check the guess
        if guess in word:
            print("Correct guess!")
        else:
            wrong_guesses += 1
            print("Wrong guess!")
            print(
                f"Attempts remaining: "
                f"{MAX_ATTEMPTS - wrong_guesses}"
            )

    print("\n" + "=" * 45)
    print("GAME OVER")
    print("=" * 45)
    print(f"The correct word was: {word.upper()}")


def main():

    while True:

        play_game()

        while True:

            choice = input(
                "\nDo you want to play again? (yes/no): "
            ).strip().lower()

            if choice in ("yes", "y"):
                break

            elif choice in ("no", "n"):
                print("\nThanks for playing!")
                return

            else:
                print("Please enter yes or no.")


if __name__ == "__main__":
    main()
```
